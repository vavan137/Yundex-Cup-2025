```python
import pandas as pd
```

```python
f = pd.read_excel('/content/C open_dataset.xlsx',index_col=None)
lines = f['name']
#print(f)
ans = []
for j in range(len(f)):
  lines = str(f['name'].iloc[j])+' '
  lines = lines.replace('а','a')
  lines = lines.replace('А','A')
  lines = lines.replace('р','p')
  lines = lines.replace('Р','P')
  #print(lines)
  only_upper = "".join([char for char in lines if char.isupper()])
  only_alpha = "".join([char for char in lines if char.isalpha()])
  codes = [ord(char) for char in only_alpha]

  if j in (40,50,114,220,261,275,381,706,913):
    print(lines+' '+str(j+1))
    #print(only_upper)
    #print(only_alpha.lower())
    #print(codes)
  if 'шaфpиaн' in only_upper.lower():
    ans.append(j+1)
  elif 'шaфp1aн' in lines.lower():
    ans.append(j+1)
  elif 'шaфpиaн' in only_alpha.lower():
    ans.append(j+1)
  elif 'shafrian' in only_alpha.lower():
    ans.append(j+1)
  elif 'shafriyan' in only_alpha.lower():
    ans.append(j+1)
print('\n'.join(map(str,ans)))

```
